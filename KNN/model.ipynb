```python
from sklearn.neighbors import KNeighborsClassifier

# Matriks fitur
X_train = [
	[0.1, -0.3],
	[-0.2, 1.5],
	[1, 3],
	[-0.6, 1.3],
	[2.3, -4.7],
	[0.4, 0.9],
	[-0.1, 1.3],
	# ...
]

# Vektor label
y_train = [
	0,
	1,
	0,
	1,
	0,
	1,
	0,
	# ...
]

# Data uji
X_test = [
    [0.1, 0.5],
    # ...
]

# Jumlah tetangga (k)
k = 5

knn = KNeighborsClassifier(
    n_neighbors=k,      # Jumlah tetangga
    weights='distance',  # Pembobotan jarak, 1/distance
    metric='minkowski', # Metrik jarak
    p=2,                # Nilai pangkat (untuk metrik minkowski)
)
knn.fit(X_train, y_train)
pred = knn.predict(X_test)
print(pred)

'''
Hyper parameter satu lagi adalah algorithm untuk opitimisasi
'''
```
